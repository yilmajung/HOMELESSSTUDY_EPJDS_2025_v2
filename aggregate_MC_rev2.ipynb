```python
# Monte Carlo aggregation to city‐daily totals with P(Y>0) thresholding
S          = 500
p_thresh   = 0.68   # drop any box whose P(Y>0) < 0.69
daily_out  = []
n_days     = df_combined_noclamp['timestamp'].nunique()

# z‐score for 95% CI
z95 = 1.96
z90 = 1.645

from tqdm import tqdm

# additions inside your loop (store city-level lambda draws for NLPD
lambda_draws_by_day = {}  # timestamp -> np.array of Λ_s

for day, grp in tqdm(df_combined_noclamp.groupby('timestamp'),
                     total=n_days,
                     desc="Aggregating city totals"):
    
    # pull out your per-box summaries
    med   = grp['median'].values    # median λ_i
    l95   = grp['lower95'].values
    # compute P(Y_i > 0) ≈ 1 - exp(-med)
    p_any = 1 - np.exp(-med)
    
    # keep only boxes likely to have tents
    keep_mask = p_any >= p_thresh
    if not keep_mask.any():
        # no boxes pass the threshold → zero tents
        daily_out.append({
            'timestamp':  day,
            'mean_total': 0.0,
            'median_total': 0.0,
            'lower95': 0.0,
            'upper95': 0.0,
            'lower90': 0.0,
            'upper90': 0.0,
        })
        continue
    
    med_filt = med[keep_mask]
    l95_filt = l95[keep_mask]
    nbox     = med_filt.size
    
    # reconstruct log-normal latent params
    mu_f    = np.log(med_filt)
    sigma_f = (np.log(med_filt) - np.log(l95_filt)) / z95
    
    # Monte Carlo draws
    f_samps = np.random.normal(
        loc=mu_f[None, :],
        scale=sigma_f[None, :],
        size=(S, nbox)
    )
    lam_samps = np.exp(f_samps)
    # optional hard cap
    lam_samps = np.minimum(lam_samps, 1e3)

    # draw Poisson counts and sum
    y_samps     = np.random.poisson(lam_samps)
    city_samps  = y_samps.sum(axis=1)

    # Store city-level lambda draws (Λ_s = sum_i λ_{i,s}) for NLPD
    lambda_draws_by_day[day] = lam_samps.sum(axis=1)

    # store summary
    daily_out.append({
        'timestamp':    day,
        'mean_total':   city_samps.mean(),
        'median_total': np.median(city_samps),
        'lower95':      np.percentile(city_samps, 2.5),
        'upper95':      np.percentile(city_samps, 97.5),
        'lower90':      np.percentile(city_samps, 5.0),
        'upper90':      np.percentile(city_samps, 95.0),
    })

# assemble final daily DataFrame
df_daily_69_noclamp = (
    pd.DataFrame(daily_out)
      .sort_values('timestamp')
      .reset_index(drop=True)
)
print(df_daily_69_noclamp.head())

# Make sure the timestamp is in datetime format
df_daily_69_noclamp['timestamp'] = pd.to_datetime(df_daily_69_noclamp['timestamp'])
df_sf['timestamp'] = pd.to_datetime(df_sf['timestamp'])

# Align by timestamp (use the ground-truth date restriction you intended)
df_eval = (
    df_sf[df_sf['timestamp'] < '2024-06-01'][['timestamp','tent_struc']]
    .merge(df_daily_69_noclamp[['timestamp','mean_total']], on='timestamp', how='inner')
    .rename(columns={'tent_struc':'y','mean_total':'yhat'})
    .sort_values('timestamp')
    .reset_index(drop=True)
)

y    = df_eval['y'].to_numpy(dtype=float)
yhat = df_eval['yhat'].to_numpy(dtype=float)

# Metrics
EPS = 1e-12

# RMSE
rmse = np.sqrt(np.mean((yhat - y)**2))

# 1) MAPE (you said no zeros in y)
mape = np.mean(np.abs(yhat - y) / (y + EPS)) * 100.0

# 2) Mean Poisson Deviance (count-likelihood grounded)
#    D = (2/n) * sum( y*log(y/mu) - (y - mu) )
def mean_poisson_deviance(y, mu, eps=EPS):
    y  = np.asarray(y, dtype=float)
    mu = np.asarray(mu, dtype=float)
    return 2.0 * np.mean(y * (np.log((y + eps) / (mu + eps))) - (y - mu))

mpd = mean_poisson_deviance(y, yhat)

# 3) NLPD using a mixture-of-Poissons predictive:
#    For each day t, p(y_t) = (1/S) * sum_s PoissonPMF(y_t | Λ_{t,s})
from scipy.special import gammaln

def log_pois_pmf(k, lam, eps=EPS):
    # vectorized: k is scalar, lam is array
    # log p(k|lam) = k*log(lam) - lam - log(k!)
    return k*np.log(lam + eps) - lam - gammaln(k + 1.0)

def day_log_predictive(k, lambda_draws):
    # log mean of exponentiated log pmfs = log( mean_s exp(log p_s) )
    # use log-sum-exp for stability
    lp = log_pois_pmf(k, lambda_draws)
    m  = np.max(lp)
    return m + np.log(np.mean(np.exp(lp - m) + 0.0))

# Build per-day NLPD by pulling the correct Λ_draws for each timestamp
logps = []
for ts, k in zip(df_eval['timestamp'].tolist(), y.astype(int)):
    lam_draws = lambda_draws_by_day.get(ts, None)
    if lam_draws is None or len(lam_draws) == 0:
        # Fallback: use single Poisson with mean yhat when draws are missing
        lam_draws = np.array([df_eval.loc[df_eval['timestamp']==ts, 'yhat'].item()], dtype=float)
    logps.append(day_log_predictive(k, lam_draws))

nlpd = -np.mean(logps)

print(f"Lambda threshold: {p_thresh}")
print(f"RMSE: {rmse:.3f}")
print(f"MAPE (%): {mape:.2f}")
print(f"Mean Poisson Deviance: {mpd:.4f}")
print(f"NLPD: {nlpd:.4f}")
```

```python
# Calculate RMSE across a range of P(Y>0) thresholds from 0.3 to 0.92

from tqdm import tqdm


# Config
S = 500
P_LIST = np.round(np.arange(0.30, 0.93, 0.02), 2)  # 0.30..0.92
z95 = 1.96
EPS = 1e-12
np.random.seed(42)

LAM_CAP_CELL = 1e4     # cap per-cell λ draw
LAM_CAP_CITY = 1e7     # cap per-city λ (sum across cells)
z95 = 1.96
EPS = 1e-12

def poisson_rng_safe(lam_vec, cutoff=1e6):
    """Vectorized RNG: Poisson for lam<cutoff; Normal approx for large lam."""
    lam = np.asarray(lam_vec, dtype=np.float64)
    lam = np.nan_to_num(lam, nan=0.0, posinf=cutoff*10, neginf=0.0)
    out = np.empty_like(lam)
    small = lam < cutoff
    out[small] = np.random.poisson(lam[small])
    big = ~small
    if np.any(big):
        mu = lam[big]
        out[big] = np.random.normal(mu, np.sqrt(mu)).round().clip(min=0)
    return out.astype(np.int64)


# Ensure datetime
df_combined_noclamp = df_combined_noclamp.copy()
df_combined_noclamp['timestamp'] = pd.to_datetime(df_combined_noclamp['timestamp'])
df_sf = df_sf.copy()
df_sf['timestamp'] = pd.to_datetime(df_sf['timestamp'])

# Containers for per-threshold outputs
daily_out_by_p = {p: [] for p in P_LIST}
lambda_draws_by_day_by_p = {p: {} for p in P_LIST}

# Group once; reuse the SAME MC draws for all thresholds
n_days = df_combined_noclamp['timestamp'].nunique()
for day, grp in tqdm(df_combined_noclamp.groupby('timestamp'),
                     total=n_days,
                     desc="Aggregating city totals (shared MC)"):
    
    med = grp['median'].to_numpy(dtype=np.float64)
    l95 = grp['lower95'].to_numpy(dtype=np.float64)

    # Guard inputs
    med = np.clip(np.nan_to_num(med, nan=0.0, posinf=LAM_CAP_CELL), EPS, None)
    # enforce l95 <= med (strict) and >0
    l95 = np.clip(np.nan_to_num(l95, nan=EPS, posinf=med), EPS, med - 1e-9)

    # Reconstruct log-normal params
    mu_f    = np.log(med)
    sigma_f = (np.log(med) - np.log(l95)) / z95
    sigma_f = np.clip(np.nan_to_num(sigma_f, nan=0.0, posinf=10.0, neginf=0.0), 0.0, 10.0)

    # Shared MC draws of latent f and λ
    S = 500
    f_samps = np.random.normal(loc=mu_f[None, :], scale=sigma_f[None, :], size=(S, med.size))
    # clip f to keep λ in a safe range
    f_samps = np.clip(f_samps, np.log(EPS), np.log(LAM_CAP_CELL))
    lam_samps = np.exp(f_samps)  # shape (S, nbox)

    # P(Y>0) from median rate for threshold gating
    p_any = 1.0 - np.exp(-med)

    # ---- if you use a single p_thresh ----
    keep_mask = p_any >= p_thresh
    if not np.any(keep_mask):
        daily_out.append({
            'timestamp':  day,
            'mean_total': 0.0, 'median_total': 0.0,
            'lower95': 0.0, 'upper95': 0.0, 'lower90': 0.0, 'upper90': 0.0,
        })
        lambda_draws_by_day[day] = np.zeros(S, dtype=float)
    else:
        # City rate per draw = sum of kept λ; clip to safe max
        lam_city = lam_samps[:, keep_mask].sum(axis=1)
        lam_city = np.clip(np.nan_to_num(lam_city, nan=0.0, posinf=LAM_CAP_CITY), 0.0, LAM_CAP_CITY)

        # Sample city counts safely
        city_samps = poisson_rng_safe(lam_city, cutoff=1e6)

        # Store for NLPD
        lambda_draws_by_day[day] = lam_city

        daily_out.append({
            'timestamp':    day,
            'mean_total':   float(city_samps.mean()),
            'median_total': float(np.median(city_samps)),
            'lower95':      float(np.percentile(city_samps, 2.5)),
            'upper95':      float(np.percentile(city_samps, 97.5)),
            'lower90':      float(np.percentile(city_samps, 5.0)),
            'upper90':      float(np.percentile(city_samps, 95.0)),
        })

        continue

    # Reconstruct log-normal params for latent f
    mu_f    = np.log(np.clip(med, EPS, None))
    # guard for l95<=0 or >=med (clip to be <= med)
    l95_adj = np.clip(l95, EPS, med - 1e-9)
    sigma_f = (np.log(np.clip(med, EPS, None)) - np.log(l95_adj)) / z95
    sigma_f = np.clip(sigma_f, 0.0, None)

    # Shared MC draws for ALL thresholds on this day
    f_samps   = np.random.normal(loc=mu_f[None, :], scale=sigma_f[None, :], size=(S, nbox))
    lam_samps = np.exp(f_samps)
    lam_samps = np.minimum(lam_samps, 1e3)  # optional cap
    y_samps   = np.random.poisson(lam_samps)  # shape (S, nbox)

    # Precompute detection probs for masking
    p_any = 1.0 - np.exp(-np.clip(med, 0, None))  # P(Y>0) under median λ

    # Apply all thresholds via masks
    for p in P_LIST:
        keep = p_any >= p
        if not np.any(keep):
            daily_out_by_p[p].append({
                'timestamp':  day,
                'mean_total': 0.0,
                'median_total': 0.0,
                'lower95': 0.0,
                'upper95': 0.0,
                'lower90': 0.0,
                'upper90': 0.0,
                'active_boxes': 0,
            })
            lambda_draws_by_day_by_p[p][day] = np.zeros(S, dtype=float)
            continue

        # Sum only included boxes
        city_samps = y_samps[:, keep].sum(axis=1)              # mixture-of-Poisson counts
        lam_sum    = lam_samps[:, keep].sum(axis=1)            # Λ_s for NLPD etc.

        daily_out_by_p[p].append({
            'timestamp':   day,
            'mean_total':  float(city_samps.mean()),
            'median_total': float(np.median(city_samps)),
            'lower95':     float(np.percentile(city_samps, 2.5)),
            'upper95':     float(np.percentile(city_samps, 97.5)),
            'lower90':     float(np.percentile(city_samps, 5.0)),
            'upper90':     float(np.percentile(city_samps, 95.0)),
            'active_boxes': int(keep.sum()),
        })
        lambda_draws_by_day_by_p[p][day] = lam_sum


# Build per-threshold daily frames & evaluate RMSE

def mean_poisson_deviance(y, mu, eps=EPS):
    y  = np.asarray(y, dtype=float)
    mu = np.asarray(mu, dtype=float)
    return 2.0 * np.mean(y * (np.log((y + eps) / (mu + eps))) - (y - mu))

results = []
daily_frames = {}

# Ground-truth restriction as in your code (pre-June 2024)
sf_eval = df_sf[df_sf['timestamp'] < '2024-06-01'][['timestamp','tent_struc']].rename(columns={'tent_struc':'y'})

for p in P_LIST:
    df_daily = pd.DataFrame(daily_out_by_p[p]).sort_values('timestamp').reset_index(drop=True)
    df_daily['timestamp'] = pd.to_datetime(df_daily['timestamp'])

    df_eval = (
        sf_eval.merge(df_daily[['timestamp','mean_total']], on='timestamp', how='inner')
               .rename(columns={'mean_total':'yhat'})
               .sort_values('timestamp')
               .reset_index(drop=True)
    )

    if len(df_eval) == 0:
        results.append({'p_thresh': p, 'RMSE': np.nan, 'MAPE(%)': np.nan, 'MeanPoissDev': np.nan})
        continue

    y    = df_eval['y'].to_numpy(dtype=float)
    yhat = df_eval['yhat'].to_numpy(dtype=float)

    rmse = float(np.sqrt(np.mean((yhat - y)**2)))
    mape = float(np.mean(np.abs(yhat - y) / (y + EPS)) * 100.0)
    mpd  = float(mean_poisson_deviance(y, yhat))

    results.append({'p_thresh': p, 'RMSE': rmse, 'MAPE(%)': mape, 'MeanPoissDev': mpd})
    daily_frames[p] = df_daily

# Summary table across thresholds
df_rmse_sweep = pd.DataFrame(results).sort_values('p_thresh').reset_index(drop=True)
print(df_rmse_sweep)

# Example: best p by RMSE
best_row = df_rmse_sweep.loc[df_rmse_sweep['RMSE'].idxmin()]
print(f"\nBest p_thresh by RMSE: {best_row['p_thresh']:.2f} (RMSE={best_row['RMSE']:.3f})")

# Optionally save
# df_rmse_sweep.to_csv("stvgp_rmse_sweep.csv", index=False)
# For a specific p (e.g., 0.70) you can access daily frame via: daily_frames[0.70]
```

### Smoothing

```python
import datetime as dt
import statsmodels.api as sm

lowess_smoothed = sm.nonparametric.lowess(
    df_merge_69_noclamp['mean_total'],
    df_merge_69_noclamp['timestamp'].astype(int) / 10**9,
    frac=0.1,
    return_sorted=True
)

# Smooth uncertainty bands
df_merge_69_noclamp['lower95_smoothed'] = sm.nonparametric.lowess(
    df_merge_69_noclamp['lower95'],
    df_merge_69_noclamp['timestamp'].astype(int) / 10**9,
    frac=0.1,
    return_sorted=False
)

df_merge_69_noclamp['upper95_smoothed'] = sm.nonparametric.lowess(
    df_merge_69_noclamp['upper95'],
    df_merge_69_noclamp['timestamp'].astype(int) / 10**9,
    frac=0.1,
    return_sorted=False
)

df_merge_69_noclamp['smoothed_count'] = lowess_smoothed[:, 1]

start_date = dt.datetime(2019, 4, 1)
end_date = dt.datetime(2024, 5, 31)

# Plot time-series
plt.figure(figsize=(12, 6))
plt.style.use('seaborn-v0_8-whitegrid')

# Predicted mean
plt.plot(
    df_merge_69_noclamp['timestamp'], 
    df_merge_69_noclamp['smoothed_count'], 
    label='Predicted Tent Counts', 
    color='#1FA7F0',
    linewidth=1
)

plt.scatter(
    df_merge_69_noclamp['timestamp'], 
    df_merge_69_noclamp['tent_struc'], 
    label='SF Quarterly Tent Counts', 
    alpha=0.6, 
    marker='o',
    color='red'
)

# Uncertainty band
plt.fill_between(
    df_merge_69_noclamp['timestamp'],
    df_merge_69_noclamp['lower95_smoothed'],
    df_merge_69_noclamp['upper95_smoothed'],
    color='gray',
    alpha=0.3,
    label='95% CI'
)

plt.xlabel('Date', fontsize=14)
plt.ylabel('Number of Tents', fontsize=14)
plt.legend(fontsize=12)
#plt.ylim(0,100)
plt.xlim(start_date, end_date)

plt.tight_layout()
plt.show()
```

```python
# Max-min normalization of counts by year
def min_max_normalize(series):
    return (series - series.min()) / (series.max() - series.min())

gdf_yearly_counts['normalized_count'] = gdf_yearly_counts.groupby('year')['count_sum'].transform(min_max_normalize)
gdf_yearly_counts['normalized_count_entire'] = gdf_yearly_counts['count_sum'].transform(min_max_normalize)
gdf_yearly_counts['normalized_mean'] = gdf_yearly_counts.groupby('year')['count_mean'].transform(min_max_normalize)
gdf_yearly_counts['normalized_mean_entire'] = gdf_yearly_counts['count_mean'].transform(min_max_normalize)

# Standardize counts by year (z-score normalization)
def z_score_normalize(series):
    return (series - series.mean()) / series.std()

gdf_yearly_counts['z_score'] = gdf_yearly_counts.groupby('year')['count_sum'].transform(z_score_normalize)
gdf_yearly_counts['z_score_entire'] = (gdf_yearly_counts['count_sum'] - gdf_yearly_counts['count_sum'].mean()) / gdf_yearly_counts['count_sum'].std()
gdf_yearly_counts['z_score_mean'] = gdf_yearly_counts.groupby('year')['count_mean'].transform(z_score_normalize)
gdf_yearly_counts['z_score_mean_entire'] = gdf_yearly_counts['count_mean'].transform(z_score_normalize)
gdf_yearly_counts.head(10)
```
